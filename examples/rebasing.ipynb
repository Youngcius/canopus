```python
from scipy.stats import unitary_group
import canopus
from qiskit.quantum_info import Operator
import numpy as np
from qiskit.circuit.random import random_circuit
```

### B gate optimal decomposition and ISA rebase

```python
# target requires 0 B gates
u = np.kron(unitary_group.rvs(2), unitary_group.rvs(2))
qc = canopus.decomposition.two_qubit_unitary_to_b_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires 1 B gates
u = canopus.BGate().to_matrix()
qc = canopus.decomposition.two_qubit_unitary_to_b_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires 2 B gates
u = unitary_group.rvs(4)
qc = canopus.decomposition.two_qubit_unitary_to_b_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires 2 B gates
u = unitary_group.rvs(4)
```

```python
qc = random_circuit(5, 100, max_operands=2)
qc_b = canopus.rebase_to_b(qc)
# assert Operator(qc_b).equiv(qc)
print('infidelity:', canopus.utils.infidelity(canopus.utils.qc2mat(qc), canopus.utils.qc2mat(qc_b)))
qc_b.draw(fold=-1)
```

### √iSWAP gate optimal decomposition and ISA rebase

```python
def sandwich_single_qubit_unitaries(u):
    return np.kron(unitary_group.rvs(2), unitary_group.rvs(2)) @ u @ np.kron(unitary_group.rvs(2), unitary_group.rvs(2))

u = unitary_group.rvs(4)
```

```python
# target requires 0 √iSWAP gates
u = sandwich_single_qubit_unitaries(np.identity(4))
qc = canopus.decomposition.two_qubit_unitary_to_sqisw_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires 1 √iSWAP gates
u = sandwich_single_qubit_unitaries(canopus.CanonicalGate(0.25, 0.25, 0).to_matrix())
qc = canopus.decomposition.two_qubit_unitary_to_sqisw_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires 2 √iSWAP gates
u = sandwich_single_qubit_unitaries(canopus.CanonicalGate(0.4, 0.3, 0).to_matrix())
qc = canopus.decomposition.two_qubit_unitary_to_sqisw_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires 3 √iSWAP gates
u = sandwich_single_qubit_unitaries(canopus.CanonicalGate(0.5, 0.3, 0.25).to_matrix())
qc = canopus.decomposition.two_qubit_unitary_to_sqisw_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# end-to-end circuit rebase
qc = random_circuit(5, 30, max_operands=2)
qc_sqisw = canopus.rebase_to_sqisw(qc)
assert Operator(qc_sqisw).equiv(qc)
qc_sqisw.draw(fold=-1)
```

### Canonical gate optimal decomposition and ISA rebase

```python
# target requires non-entanglement 4x4 unitary
u = sandwich_single_qubit_unitaries(np.identity(4))
qc = canopus.decomposition.two_qubit_unitary_to_canonical_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# target requires random 4x4 unitary
u = unitary_group.rvs(4)
qc = canopus.decomposition.two_qubit_unitary_to_canonical_circuit(u)
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# end-to-end circuit rebase
qc = random_circuit(4, 50, max_operands=2)
qc_can = canopus.rebase_to_canonical(qc)
assert Operator(qc_can).equiv(qc)
qc_can.draw(fold=-1)
```

### Customized gate set optimal decomposition and ISA rebase

```python
# target requires random 4x4 unitary
u = unitary_group.rvs(4)
qc = canopus.decomposition.two_qubit_unitary_to_custom_circuit(
    u, gate_set=[canopus.CanonicalGate(0.25, 0, 0), canopus.CanonicalGate(0.5, 0.3, 0)], costs=[0.8, 1.0])
assert Operator(qc).reverse_qargs().equiv(u)
qc.draw(fold=-1)
```

```python
# end-to-end circuit rebase (now built on Gulps, with numerical precision typically ~ from 1e-10 to 1e-13)
gate_set = [canopus.CanonicalGate(0.25, 0, 0), canopus.CanonicalGate(0.5, 0.3, 0)]
qc = random_circuit(4, 15, max_operands=2)
qc_custom = canopus.rebase_to_custom(qc, gate_set=gate_set, costs=[0.8, 1.0], names=['Can', 'Can'])
print('infidelity:', canopus.utils.infidelity(canopus.utils.qc2mat(qc), canopus.utils.qc2mat(qc_custom)))
qc_custom.draw(fold=-1)
```

```python
canopus.utils.print_circ_info(qc_custom)
```
